# Stage 3: Advanced RAG — Agentic and Graph RAG

In Stage 3, we move past basic context extraction into **Multi-Document Reasoning**, **Agentic Loops**, **Graph RAG**, and automated **RAGAS evaluation**.

### Learning Objectives
1. **Graph RAG (Neo4j)**: Model connections between contracts (entities, signing parties, dates, restrictions).
2. **Agentic Loops (LangGraph)**: Set up a state machine that audits contracts, checks for document-level conflicts, and drafts audit logs.
3. **Missing Clause Detector**: A proactive agent that inspects what is *not* present in the files.
4. **Evaluation**: Benchmark retrieval accuracy using the RAGAS framework.

```python
# 1. Load Environment & Imports
from dotenv import load_dotenv
from src.agent import LegalDueDiligenceAgent

load_dotenv()
print("Stage 3 modules loaded.")
```

```python
# 2. Neo4j Graph Model Definition (Conceptual schema)
schema_info = """
Graph Database Schema:
  (:Party {name: 'Founder'}) -[:SIGNS]-> (:Agreement {name: 'Employment Contract'})
  (:Agreement) -[:CONTAINS]-> (:Clause {id: 'IP_Clause_01', type: 'IP_Assignment'})
  (:Clause) -[:CONTRADICTS]-> (:Clause {id: 'NDA_IP_Clause', type: 'Confidentiality'})
"""
print(schema_info)
```

```python
# 3. Run the LangGraph-style Audit Agent Loop
agent = LegalDueDiligenceAgent()

# Define documents to send to the auditor
document_paths = [
    "data/raw/Mutual NDA.pdf",
    "data/raw/Employment Agreement.pdf",
    "data/raw/TERM_SHEET_EQUITY.pdf"
]

result_state = agent.run_agent_loop(document_paths)
print("Audit State execution finished.")
```
